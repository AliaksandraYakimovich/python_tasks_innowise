**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
# your code

A = np.array([
    [1, 2, 3],
    [7, 8, 9],
    [3, 4, 5],
    [1, 1, 6]
])

B = np.array([
    [1, 3],
    [4, 6]
])

Array = A[np.all([np.isin(A, b_row).any(axis=1) for b_row in B], axis=0)]
print(Array)
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
# your code

from numpy import matrix

matrix = np.array([[1, 2, 3], [4, 5, 6], [1, 2, 3], [7, 8, 9], [4, 5, 6], [1, 1, 1], [2, 2, 2], [3, 3, 3], [4, 4, 4], [5, 5, 5]])
result = matrix[matrix.max(axis=1) != matrix.min(axis=1)]
print("Matrix with unique elements within a row:")
print(result)
```

**6.** Дан двумерный массив. Удалить те строки, которые повторяются

```python
# your code

Array = np.array([
    [7, 8],
    [1, 2],
    [4, 5],
    [1, 2]  
    ])

k, indices = np.unique(Array, axis=0, return_index=True)
unique_arr = Array[np.sort(indices)]
print("Array without duplicate rows:")
print(unique_arr)
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль. 
 Например, для x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) ответ 5.

```python
# your code

x = [6, 2, 0, 3, 0, 0, 5, 7, 0]

next_ind0 = [x[i + 1] for i in range(len(x) - 1) if x[i] == 0]
print(max(next_ind0))
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
# your code

x = [2, 2, 2, 3, 3, 3, 5]

values = []
counts = []
for i in x:
    if not values or i != values[-1]:
        values.append(i)
        counts.append(1)
    else:
        counts[-1] += 1
print(values, counts)
```

```python

x = np.array([2, 2, 2, 3, 3, 3, 5])

groups = np.split(x, np.where(np.diff(x) != 0)[0] + 1)
values = np.array([g[0] for g in groups])
counts = np.array([len(g) for g in groups])
print(values, counts)
```
